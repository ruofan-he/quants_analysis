モンテカルロ計算を行うエンジンをpythonで作りたい。
それぞれクラス分けして生成するパスを自由に拡張したり、商品を自由に拡張できる設計にしたい。

環境クラス : シミュレーショングリッド、およびシミュレートする乱数のシードを管理するクラス
SDEクラス : HWモデルでの金利のSDE(国内金利のもの、これのmoney-marketアカウントを全てのSDEの基準材とする。) HWモデルでの外国金利のSDE, HW-BSモデルでの株価のSDE
pathクラス : 金利pathクラス,株価pathクラスなどシミュレーション対象となるSDEに従って生成するパスを指定するクラス
indexクラス : 生成したpathから商品をシミュレートするために必要な参照値を決めるクラス。
payoffクラス : 生成したindexクラスを指定して商品のペイオフを指定するためのクラス。

これらをもとにデリバティブの計算を行うエンジンを作りたい。
